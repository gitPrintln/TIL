## Quantum Gate
- 단순한 회로 명령이 아니라, 데이터를 새로운 표현으로 변환하는 Feature Transformation

## 대표 Quantum Gate
1. X Gate은 큐비트의 상태를 반전시키는 게이트
- 0 -> 1, 1 -> 0(0과 1을 교환)
2. Y Gate
- Y 축을 기준으로 회전시키며, 값 변경과 '위상' 변화까지 일어나는 게이트
3. Z Gate
- 위상 반전만 일으킴 값은 그대로임 앞에 - 붙임(0은 그대로, 1은 -1)
4. H Gate
- 중첩 상태 생성, 0 1을 확장(확률 분산)
5. CNOT Gate
- 얽힘 상태 생성

## Quantum Gate는 어떤 결과가 나오는가 보다 상태를 어떻게 변화시키는가 이해하는 것이 중요하다.
