## counts 는 측정 결과의 빈도일 뿐, 큐비트의 상태 자체를 직접 보여주는 것은 아님.
## counts -> shot 으로 나누기 전임 = 빈도(횟수)
## 신뢰할 수 있는 해석을 위해 충분한 shot tnfmf tkdydgodigka.

## counts 만 보고 상태를 단정할 수 없다.
-> 측정 전 큐비트가 어떤 '양자 상태'였는지는 counts 만으로는 알 수 없다.
-> counts 는 양자 출력의 '정량적 표현'
-> counts = 양자 상태의 확률 분포를 수치(데이터)로 표현한 것.
-> 머신 러닝 모델의 입력 특징이나 예측 타깃으로 활용할 수 있다.

## Qiskit 는 기본적으로 리틀 엔디안(little Endian)임

## histogram 은 측정 결과가 얼마나 자주 나타났는지를 막대그래프로 표현한 것.
-> counts를 시각적으로 표현한 것.
-> 두 가지 서로 같은 정보를 다르게 표현
-> 초기상태 에서 아무 게이트도 적용하지 않으면 상태는 변하지 않는다. : 따라서 측정해도 항상 0만 나온다.
-> H Gate 를 적용시 측정하면 '0'과 '1' 이 비슷한 분포 50:50 부근 이 나온다.
-> X Gate 적용시 측정하면 '1'만 100% 나온다.

방법 : plot_histogram() 함수를 사용

## counts 는 데이터, histogram 은 데이터 요약과 해석 도구이다.
