## correlation: Quantum Entanglement를 이해하려면 상관관계를 이해하는 것이 중요하다. (기초단계)
- 상관 관계 란? 두 개 이상의 대상이나 변수 사이에 일정한 관련성이 존재하는 관계를 의미한다.
-> 한 변수의 변화가 다른 변수의 변화와 함께 나타나는 현상
-> key point: 원인과 결과가 아님, '함께 변하는 경향'일 뿐
ex) 기온과 아이스크림 판매량, 공부시간과 시험 점수, 운동량과 체중(Negativ Correlation)

## classical correlation 의 특징
1. 각각의 값은 측정하기 전부터 이미 정해져 있다.
2. 각각의 값은 독립적으로 설명할 수 있다. 따로 떼어 설명, 개별 측정 및 저장 가능
3. 인과관계와 다르다. -> 원인과 결과 관계가 아님


## classical correlation vs quantum entanglement
-> 두 상관관계 모두 '연관성'은 있지만 그 성질과 원리는 완전히 다르다.
-> classical correlation: 원인 결과(정보 필요), quantum entanglement: 연결 결과(정보 불필요)
