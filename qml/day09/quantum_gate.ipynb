## Quantum Gate 조합
- Gate 조합이란 하나의 큐비트에 여러 Gate를 순서대로 적용하는 것이다.
예시) H -> X : 먼저 H Gate 를 적용하고, 그 다음 X Gate 를 적용
-> 순서대로 적용됨

## 상태 변화 
-> 측정하면 0과 1이 거의 같은 확률로 관측될 것으로 예상된다.
## 결과 분석
같은 히스토그램이 나와도 Gate 조합은 큐비트의 내부상태를 변화시킨다.

## 순서의 중요성
1. H -> X
2. X -> H

-> Gate 순서를 바꾸면 상태변화과정이 달라지고 최종 Quantum State(위상)도 달라진다.
-> Histogram 이 같아도 Quantum State 는 다를 수 있다.
-> 히스토그램만으로 차이를 알 수 없을 수 있다.

## Quantum Gate 별 결과 비교가 필요한 이유
-> 

## Gate 결과 분석 : 1. 결과는 State의 직접 표현이 아니다. = 우리가 보는 결과는 큐비트 State 그 자체가 아님, 측정되기 전까지 연속적인 가능성으로 존재함

## Gate 결과 분석 : 2. 같은 히스토그램이라도 State가 다를 수 있다. = 히스토그램만으로 큐비트 상태를 완전히 알 수 없다.(방향(위상정보) 는 표현 되지 않음) => 더 많은 Gate 조합/ 다양한 측정이 필요함

## Gate 결과 분석 : 3. Quantum Gate 는 결과가 아니라 상태를 바꾼다. =  Gate 를 적용해도 바로 0 또는 1이 나오는 것이 아니다. 측정할 때 비로소 0 또는 1 결과(확률분포) 로 나온다.
