## feature transformation 은 모델에 더 좋은 입력을 만들어주기 위해 기존의 feature 을 새로운 표현으로 바꾸는 과정

예시1) 제곱 변환 : Feature 값을 제곱하여 새로운 feature 를 만드는 방법
예시2) 로그 변환 : Feature 값의 범위를 압축하여 큰 값의 영향을 줄이고, 더 안정적인 학습을 가능하게 한다.
예시3) 비율 Feature 생성 : 두 Feature의 비율을 사용하면, 절대 값보다 상대적인 관계를 더 잘 표현할 수 있다.
예시4) 상호 작용 Feature 생성 : 두 Feature 곱하여 두 변수의 조합 효과(상호작용)를 표현하는 새로운 Feautre를 만든다.

결국 목적은 모델이 데이터를 더 잘 이해하고, 더 좋은 성능을 내도록 돕는것
1. 비선형 관계를 선형적으로 변환
2. 스케일(크기) 차이를 줄여 안정적인 학습
3. 이상치(Outlier)의 영향을 줄임
4. 특징 간 관계(상호작용)를 더 잘 표현
5. 모델의 성능(정확도)를 향상

## Quantum Encoding 이 적용 되면, 데이터는 0 상태에서 새로운 양자 상태로 변환한다.
-> Encoding 은 데이터를 양자 상태 벡터로 변환한다.
-> 변환된 상태 x(인코딩 이후의 양자 상태) 는 Bloch Sphere의 한 점으로 표현된다.
-> 이 상태가 바로 이후 회로(Variational Circuit)와 측정을 통해 모델이 학습하고 예측하는 기반이 된다.

## Encoding 만으로 충분할까? Encoding은 시작일 뿐이다.
-> Encoding
-> 회로 연산(변환, 학습)
-> 측정(출력, 예측)
-> Encoding은 데이터를 양자 상태로 표현 하는 첫 단계
-> 위의 흐름이 필요

## Quantum Gate 적용 전후 상태 변화
-> 1. Encoding 후 상태
-> 2. Gate(회로) 적용
-> 3. 상태 변화(수학적 표현)
-> 4. 새로운 양자 상태
-> 5. 측정 시 결과 분포 변화

## zzFeatureMap
-> 단순한 회로가 아니라 입력 데이터를 Quantum Feature 로 변환하여 QML 모델이 더 잘 학습할 수 있도록 돕는 핵심 다리(Bridge) 역할을 한다.


## Quantum Gate 은 Gate 하나(레고 블록 하나) -> 여러 레고 블록 완성된 집(QNN)
-> Quantum Gate 가 모이면 Qunatum Circuit가 된다.
-> circuit은 데이터를 단계적으로 변환(Transformation) 한다.
-> Feature Map은 Quantum Gate 들의 조합으로 Quantum Circuit 이다.
-> zzFeatureMap 에 포함 된 대표적인 Gate들(H, RZ(회전), CX(Entangling), RZZ(상호작용)등)
-> Feature Map 다음에 Variational Circuit 가 온다.
-> 실제 학습이 Variational Circuit 에서 이루어지며, Featurre Map(데이터 변환), 즉 두 단계가 함께 이루어져야 QNN이 완성된다.
