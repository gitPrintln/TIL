## Variational Circuit 란?
-> Gate 안의 파라미터를 학습하는 Circuit이다.
-> optimizer가 파라미터를 반복적으로 조정하여 모델의 성능을 향상시킨다.
-> 파라미터(세타)를 바꾸면 Quantum State 가 변경된다.
