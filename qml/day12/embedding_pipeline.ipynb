```python
import pandas as pd

customer = pd.DataFrame({
    "Age":[25,32,47,52],
    "Income":[45,52,80,90],
    "VIP":[0,0,1,1]
})


import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

plt.scatter(
    customer["Age"],
    customer["Income"],
    c=customer["VIP"],
    s=120
)

plt.xlabel("Age")
plt.ylabel("Income")
plt.title("Classical Feature Space")

plt.grid(True)

from qiskit.circuit import Parameter
from qiskit import QuantumCircuit

age = Parameter("Age")
income = Parameter("Income")

qc = QuantumCircuit(2)

qc.h(0)
qc.h(1)

qc.ry(age, 0)
qc.ry(income, 1)

qc.cx(0, 1)

qc.draw("mpl")

bound_qc = qc.assign_parameters({
    age: 25,
    income: 45
})

bound_qc.draw("mpl")
```
