Quantum Advantage : 항상 보장 x, 문제에 따라 QML 이 적합한지 판단 필요
q) 어떤문제에 적용해야하는가? -> QML로 모두 바꿀 필요 x, QML을 검토할 가치는 있다.
- 판단 기준
1. 데이터가 복잡하다.
2. Feature 수가 많다.
3. 기존의 ML성능이 한계에 도달했다.
4. 경우의 수가 폭발한다.
5. 근사해도 의미가 있다.

* 위의 판단 기준 여러 조건을 성립하면 QML을 검토해 볼 만하다.

1. 적용 가치가 낮음
- 선형 회귀 문제, 단순 매출 예측, 고객 이탈 예측
2. 적용 가치가 높음
- 복잡한 분류 문제, 고차원 데이터, 최적화 문제, 복합 패턴 탐지 문제

QSVM : Quantum Support Vector Machine
-> SVM 에 양자 컴퓨팅 원리를 결합한 양자 머신러닝의 핵심 알고리즘
방식)
1. 데이터 인코딩
2. 양자 커널 계산
3. 최적화 및 분류

장점)
1. 클래식으로는 힘든 비선형 패턴 포착
2. QML 입문자용 정석
3. 복잡한 데이터 분류 가능성
4. 고차원 특징공간 활용
ex) 주로 분류 문제, 사기거래 탐지 등
단점)
1. 데이터 전송의 병목
2. 시뮬레이션 한계

QNN : 데이터 패턴을 학습한 후 예측
-> 양자 회로를 결합한 복잡한 패턴을 효과적으로 학습할 수 있는 새로운 신경망 모델
ex) 의료(신약 후보, 질병 예측), 금융(주가, 사기예측), 제조산업(센서, 품질), 이미지 분류, 영상 분류, 추천 시스템(사용자 선호도, 콘텐츠 추천), 과학 연구

한계점
1. 한계 존재 2. 인코딩의존성, 노이즈, 하드웨어

QAOA : 여러 가능한 선택지 중 가장 최적해를 찾는 것
-> 조합 최적화 문제에 특히 효과적

ex) 배송 경로 최적화, 공장 스케줄 최적화, 포트폴리오(주식) 최적화, 배차 문제, 네트워크 경로 최적화
