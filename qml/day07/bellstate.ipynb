## Bell State
-> Bell state를 만들기 위해 H Gate 가 먼저 필요함
-> q0 큐비트를 '중첩 상태'로 만든 후 CNOT Gate 를 사용해 q1 과 강한 연관관계(얽힘) 상태로 만든다.


## step1. 초기상태
-> 처음 두 큐비트는 모두 0 이다. 
-> 00상태

## step2. q0에 H Gate 적용
-> 첫 번째 큐비트 q0에 H Gate를 적용한다.
-> H Gate는 0 상태를 '0일 가능성 + 1일 가능성'으로 만든다.

## step3. CNOT Gate 적용
-> 얽힌 상태가 되면 두 가지 가능성이 '연결'된 상태가 된다. 
-> 이것이 Bell State

## step4. 최종 상태
-> 정규화된 Bell State
-> 측정하면 00 또는 11 두 결과만 나온다. 01, 10 은 안나옴

## Bell State 상태 변화의 의미
- 00 + 11 은 두 큐비트의 측정 결과가 독립적이지 않다는 것을 의미한다.

## Bell State 의 중요성
- 얽힘(Entanglement)을 가장 단순한 형태로 보여주기 때문임
- 중첩과 얽힘의 차이를 이해할 수 있다.
- 양자 상태를 다른위치로 전달
- 한 개의 큐비트 조작으로 2비트 정보를 전달
- 양자 통신에서 자원 효율성을 높여주는 대표적인 예
