## 왜 Strongly Entangling Layers 인가?
- 문제: 회전 게이트만 충분하지 않은 이유 -> 회전 게이트만 사용하는 회로는 각 큐비트가 독립적으로 회전할 뿐, 서로 정보를 주고받지 못한다.
- 머신러닝은 개별 Feature 보다 Feature 관계를 학습한다.
- Entanglement 가 없는 회로는 각 큐비트가 서로 독립적으로 동작하면서, Feature 간 관계(상호작용)를 회로 내부에서 충분히 표현하기 어렵다.
- Entanglement는 두 개 이상의 큐비트를 독립적으로 설명할 수 없게 만들고, 측정 결과가 서로 강하게 연관되도록 하는 양자 상태의 관계이다.
- 단순 Entanglement 은 연결 범위가 좁고 서로 다른 큐비트 간 상호작용이 부족함 -> Strongly Entangling 구조(넓은 상호 작용), 다양한 패턴의 Entanglement 를 Layer 별로 반복, 모든 큐비트가 다른 많은 큐비트와 상호작용할 기회 제공, 정보 전달 경로가 다양하고 짧아져 표현력 증가

## Strongly Entangling Layer 의 기본 구조
Rotation Layer(개별 표현 학습) -> Entanglement Layer(Feature 관계 학습)

## 왜 Rotation 과 Entanglement를 반복하는가? : Rotation 으로 상태를 다양하게 만들고 Entanglement 로 관계를 만들고, 이를 반복해야 복잡한 Feature 관계를 충분히 표현할 수 있다.

## 연결 범위(Range)는 얽힘이 퍼져나가는 '거리'이다. 
=> Range가 넓을 수록 더 멀리 있는 큐비트끼리 정보를 빠르게 공유가능, QNN은 Range(연결 범위)를 어떻게 설정하느냐에 따라 성능과 하드웨어 적합성이 완전히 달라진다.

## Strongly Entangling 의 장점
- 높은 표현력(High Expressivity) : 힐베르트 공간(Hilbert Space) 전체를 얼마나 넓고 고르게 탐색할 수 있는지를 뜻한다.
- 빠른 정보 확산 (Fast Information Spread / Scrambling) : 입력된 데이터나 특정 큐비트의 변화가 전체 양자 시스템에 얼마나 빠르게 퍼져나가는가에 대한 지표
- 풍부한 상관관계 (Rich Correlations) : 단순히 두 큐비트 간의 관계를 넘어 다중 큐비트 간의 고차(Higher-order) 상관관계를 포착하는 능력
- 파라미터 효율성 (Parameter Efficiency) : 회로의 복잡도 대비 적은 층(reps)으로도 강력한 성능을 내는 능력
- 유연한 탐색 공간과 대칭성 파괴 (Flexibility & Symmetry Breaking) : 특정 패턴이나 편향에 갇히지 않고 최적의 해를 찾아가는 능력
- 복잡한 확률 분포 및 생성 모델링 적합성 (Suitability for Generative Modeling) : 복잡한 데이터의 확률 분포를 그대로 모사하는 능력
-> 베르트 공간 전체에 걸쳐 복잡하고 정교한 진폭(Amplitude) 및 위상(Phase) 분포를 만들어낼 수 있기 때문에, 양자 생성적 적대 신경망(Quantum GAN)이나 복잡한 금융 데이터, 센서 데이터의 분포를 학습하는 생성 모델(Generative Model)의 안자스(Ansatz)로 매우 적합

## Strongly Entangling 가 항상 좋은 것은 아니다.
- 표현력과 학습 안정성의 Trade-off
- Barren Plateau 위험 증가
- 회로 복잡도 및 실행 비용 증가
- 문제에 따라 과적합 위험
- 데이터 규모의 불균형
- 하드웨어 제약고려 필요

## Strongly Entangling 설계시 고려사항
- Strongly Entangling 구조는 강력하지만, 무조건 깊고 촘촘한 구조가 좋은 것은 아니다.
- 문제의 특성, 하드웨어 제약, 학습 안정성 등을 고려하여 적절한 구조를 설계해야 한다.


## RealAmplitudes와 비교: RealAmplitudes 는 QNN에서 널리 사용하는 Ansatz(변분 회로)이며, 그 핵심이 바로 Strongly Entangling Layer를 반복해서 쌓는 구조이다.

## Strongly Entangling 의 의미
- Strongly Entangling 은 단순히 게이트를 많이 사용하는 것이 아니라, 여러 Layer 를 거치며 대부분의 큐비트이 다양한 Qubit 과 충분히 연결되는 구조를 의미함

## 구조
- Strongly Entangling Layer = (Rotation Layer + Entanglement Layer) X L (반복)

## RealAmplitudes 와 Strongly Entangling 비교
-> RY 회전 중심(단일 축) vs Rot(RZ-RY-RZ) 등 다축 회전
-> 고정된 단순 패턴(주로 사다리 얽힘) vs Layer마다 다양한 패턴(Circular, Full, Range) 사용가능 얽힘
-> 표현력 : 중간 vs 높음
-> 파라미터수 : 상대적 적음 vs 상대적 많음
-> 회로 깊이 : 낮음(가벼움) vs 높음(무거움)
