## 왜 Measurement 결과만으로 충분하지 않을까? 
-> Measurement 결과가 나왔는데 Target 과 비교하지 않으면 어떤 문제가 생길까요?
= Target 과 비교하여 Loss(차이)를 계산해야, 회로의 결과가 '좋은지/나쁜지' 평가할 수 있다.
= Measuremet 결과(Counts)는 확률로 변환할 수 있고, 이 확률을 Prediction(예측값)으로 사용할 수 있다.

정리: Measurement -> 확률(P(0), P(1)) -> Prediction
Prediction을 Target과 비교하여 Loss 를 계산한다.
확률은 모델의 출력값이 아니라, Prediction(예측값)으로 변환되어야 한다.

## Expectation Value 를 Prediction으로 사용하는 경우(Counts 가아니라)
-> Measurement -> <Z> {Expectation Value} -> Prediction -> Target 과 비교 -> Loss 계산

참고: Counts는 '얼마나 자주 측정되었는지' 보여주는 값이다.

## Measurement 는 확률적인 과정이기 때문에 측정마다 달라질 수 있음
## Quantum Circuit의 parameter는 아직 학습되지 않았기 때문에 원하는 Prediction이 나오지 않고, Loss도 크다.
- 파라미터는 학습 전 임의의 초기값일 뿐
- 아직 학습되지 않았기 때문에 좋은 Prediction이 나오지 않고 Loss 가 크다
## Quantum Circuit은 처음 임의의 파라미터 로 구성되어 있어서, 아직 데이터 패턴을 반영하지 못함
- 패턴을 잘 표현할 수록 원하는 Predicttion이 나오고 Loss 는 작아진다.
## 아무리 Parameter를 잘 조정해도, 회로 구조가 단순하면 데이터의 복잡한 패턴을 표현할 수 없다.
## 회로 구조가 같아도 parameter 값이 달라지면, 측정확률이 달라진다.
## Target(정답)에 가장 가까운 Prediction을 만드는 Parameter 를 만드는 과정이 바로 학습이다.
## Prediction 과 Target의 차이를 수치화해야한다.
- 왜 수치화를 해야하나?
=> 수치가 있어야 좋은지 나쁜지 판단 가능
=> 여러 Parameter를 비교할 수 있음.
=> Optimizer가 방향을 알고 Parameter를 조정할 수 있음
=> Loss(손실 함수)는 Prediction과 Target의 차이를 숫자로 나타낸 값
- 수치화 하지 않으면?
=> 비교가 어렵고, 어떤 방향으로 조정해야 할지 알수 없음.
