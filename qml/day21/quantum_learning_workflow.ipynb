## 왜 전체 workflow 로 이해해야 하는가? 
- 각 단계는 하나의 학습 시스템 안에서 연결되어 반복되며 모든 구성 요소가 함께 작동할 때 Quantum Machin Learning이 완성된다.

## Quantum Learning Pipeline 은 여러 입력을 받아 반복 학습을 수행하고 다양한 결과(출력)을 생성한다.

## QL 에서 결국 학습되는 것은 회로 구조나 게이트가 아니라 Parameterized Quantum circuit의 파라미터이다.
- 변경되는 것은 회전 게이트의 각도 값 뿐이다.
-> 비유
-> 회로 구조 = 기계의 설계도(고정)
-> 파라미터 = 값을 조정(다이얼을 돌려 값을 조정)

## 학습 전 상태는 Optimizer 가 parameter 를 변경하기 전의 초기 상태이다.
## 학습 후 상태는 Optimizer 가 Loss 를 최소화하기 위해 Parameter를 반복적으로 수정한 결과이다.

## 학습 전후 비교의 핵심 항목: 1. Parameter 비교
- 학습 전과 학습 후의 Parameter를 비교하여 Optimizer가 회로의 회전각을 어떻게 변경했는지 확인한다.

## 최적 Parameter는 반드시 하나 인가?
- 1. Loss Landscape 는 여러 개의 골짜기를 가질 수 있다.
- 2. 서로 다른 Parameter가 같은 성능을 만들 수 있다.(예: 회로의 대칭성)
- 3. Optimizer는 하나의 해를 찾아낸다.
- 4. 예시: 두 가지 다른 최적 Parameter -> 같은 Prediction과 Loss를 만든다.
