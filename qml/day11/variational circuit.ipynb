## neural network 의 weight가 학습 대상이라면, variational circuit 은 Gate의 파라미터가 학습 대상이다.

## 안자츠(Ansatz)?
- 문제를 풀기 위해 미리 정해둔 학습 가능한 회로 구조

## Feature Map 과 Variational Circuit의 차이: 
- Feature Map 은 데이터를 Quantum State로 인코딩(변환)하는 역할을 하고 Variational Circuit은 그 상태를 학습(조정)하여 더 좋은 예측을 만드는 역할을 한다.

## ansatz는 모델의 유연성을 담당한다
- 파라미터가 많을 수록 표현력이 커진다.
- 학습을 통해 최적의 파라미터 값을 찾는다.

## 왜 두 circuit를 결합하나?
-> Feature Map 과 Variational Circuit은 각각 다른 역할을 하지만, 두 circuit을 결합해야 데이터 표현+최적화(학습)을 한 번에 수행할 수 있다.

## 같은 circuit 도 보는 관점에 따라 의미가 달라짐 -> 다시 해석해야함 -> 어떤 Feature를 만들어 내는가 를 이해할 수 있다.

## Quantum Circuit은 단순히 큐비트의 상태를 바꾸는 장치가 아니라, 입력 데이터를 새로운 Feature Representation으로 변환하는 장치이다.

## Quantum State는 물리학적으로 양자 상태이지만, ML 관점에서는 Feature Representation이다.

## Quantum State에는 단순한 큐비트 상태가 아니라, 데이터(Feature)에 대한 다양한 정보를 담고 있다.
- Amplitude 정보, Phase 정보, Correlation 정보
