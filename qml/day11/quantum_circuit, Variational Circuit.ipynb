## quantum circuit 비교가 필요한 이유
- Quantum Circuit 은 Gate 의 종류, 순서, Qubit 연결 방식, Enatanglement 여부, Measurement 결과 에 따라 결과가 달라진다.
-> 한 회로만 실행해서는 이해하기 어렵다. 여러 회로를 비교해야 알 수 있다.

## 좋은 circuit?
- 단순히 Gate 가 많은 회로가 아님
- 좋은 circuit 은 목표에 맞는 상태 변화를 만들어 내는 회로이다.

- 문제마다 적합한 구조가 다르기 때문
- 자원(하드웨어) 제약을 골하기 위해
- 모델 해석과 개선을 위해

key point) 같은 입력 데이터를 넣었을 때 어떤 Quantum Circuit 가 데이터를 더 잘 구분(식별) 할 수 있을까?

## quantum circuit 표현력이란?
- 회로가 얼마나 다양하고 복잡한 Quantum State를 만들 수 있는지를 의미한다.
- 영향을 주는 요인
=> Gate 개수, Gate 순서, Quantum Circuit Depth, 얽힘 구조, 파라미터 개수

## 표현력이 높은 circuit일 수록 더 많은 가능성을 탐색, 더 풍부한 정보를 표현 가능
-> 즉 QML 모델의 성능 향상으로 이어질 수 있다.

단.
1. 항상 좋은가? -> 그렇긴 하지만 너무 높으면 오히려 학습을 어렵게 만들 수 있다.
2. circuit depth 증가 -> 더 복잡한 상태 표현 가능, 현실적인 제약으로 인해 학습 성능이 오히려 떨어질 수 있다.
3. Barren Plateau -> 표현력이 매우 높을 때 발새하는 문제,, 파라미터를 조정해도 Cost(손실) 값이 거의 변하지 않아, Gradeient 가 0 에 가까워지고 학습이 멈추는 문제
4. overfitting -> 학습 데이터에만 지나치게 맞춰져서, 새로운 데이터(보지 못한데이터)에 대해서는 성능이 떨어지는 현상

## Variational Circuit?
- Variational Circuit 은 학습 가능한 circuit

## variational circuit은 Gate 의 parameter 가 학습된다.

## paramter Quantum Gate 는 각도 값을 받아 큐비트 상태를 회전시키는 Gate이다.
-> 파라미터는 Quantum State 를 어떻게 변환할지를 결정한다.

## Variational Circuit 예시 : RY Gate
RY Gate 는 파라미터 회전각도를 받아 큐비트를 회전시키는 Gate 이다.
회전 각도 값이 정해지기 전까지 회로가 완성되지 않으며 파라미터(회전각도) 값에 따라 측정결과 값이 달라진다.

# Variational Circuit 학습 흐름은 파라미터를 반복적으로 조정하면서 Loss 를 최소화하는 과정이다.
