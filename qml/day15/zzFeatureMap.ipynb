## Feature Map - 더 높은 차원의 공간에서 구분이 쉽도록 만듦

## zz Feature Map -> Feature Interaction 표현
-> 큐비트 사이의 얽힘을 통해 입력 feature Map
-> Quantum Kernel 에 적합 -> Quantum Feature Space 로 변환한 뒤 두 데이터 간 유사도를 계산하는데 최적화된 구조
-> 구조가 직관적 -> 회로 구조가 단순하고 규칙적이어서 이해하기 쉽고 구현 및 확장이 매우 용이
-> 실습확장이 쉽다 -> 파라미터 조절만으로 다양한 실험을 쉽게 확장할 수 있어 학습과 연구 실습에 매우적합
-> QNN으로 연결가능 -> 파라미터 회로(variational circuit)와 자연스럽게 연결되어 QNN을 구성 가능

## zz 의 의미는 두 큐비트 사이의 Pauli-z 연산자 상호 작용을 의미하며, 두 Feature 사이의 관계(상호작용)을 회로에 반영한다는 의미
- Pauli-z 연산자 : 0의 위상은 그대로 1의 위상은 뒤집음. Z는 위상(Phase)을 다루는 연산자이다.

## zzFeatureMap 에서 reps 는 회로 구조를 몇 번 반복할지를 결정하는 파라미터이다.
reps 를 늘리면 회로의 깊이가 깊어지고 표현력이 향상될 수 있지만, 계산 비용, 시간, 노이즈 영향, 과적합 위험도 함께 증가한다.

## Entanglement 방식은 어떤 큐비트들을 서로 연결(zz 상호작용)할지 결정하는 방법이다.
## Entanglement 방식에 따라 Feautre 간 상호작용(상관관계)이 표현되는 범위와 강도가 달라진다.
