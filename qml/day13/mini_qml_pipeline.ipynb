## mini QML pipeline
- 양자 컴퓨팅과 머신러닝을 결합하는 QML(Quantum Machine Learning) 분야에서, 거대한 데이터나 복잡한 회로 대신 가볍고 축소된 형태(Mini)로 데이터 전처리부터 양자 커널(Quantum Kernel) 또는 양자 신경망(QNN) 학습, 평가까지의 과정을 소형화하여 구현한 테스트/실습용 파이프라인을 의미

## 구조
1. Data Preprocessing (데이터 전처리): 고차원 클래식 데이터를 양자 상태에 올리기 위해 PCA 등으로 차원을 축소 (Amplitude Encoding, Angle Encoding 등 활용)

2. Quantum Circuit / Model (양자 회로 구성): Qiskit, PennyLane 등의 라이브러리를 활용해 매개변수화된 양자 회로(PQC, Parametrized Quantum Circuit)를 정의

3. Execution & Training (학습 및 시뮬레이터 실행): AerSimulator 등을 통해 회로를 실행하고, 고전 최적화 알고리즘(COBYLA, Adam 등)과 연동하여 파라미터 업데이트

4. Evaluation (평가): 예측 성능(Accuracy, Loss 등)을 확인
