## SPSA 는 Simulationeous Perturbation Stochastic Approximation 의 약자로, 기울기를 추정하여 파라미터를 업데이트하는 Optimizer 이다.

## SPSA 가 필요한 문제 상황: 파라미터가 여러 개인 경우 각 파라미터를 하나씩 변경하며 평가하려면 필요한 Loss 평가 횟수가 매우 많아지게 된다.

## SPSA에서 Stochastic는 Loss 평가가 확률적(노이즈가 있음)이라는 의미이다. 즉 같은 파라미터에서 평가해도 매번 결과가 달라질 수 있다.

## SPSA의 Loss Curve 가 흔들리는 이유는 두 번의 Loss 평가만으로 Gradient를 근사하기 때문에 노이즈의 영향을 받기 때문이다.

## SPSA에서 Perturbation은 Gradient를 추정하기 위해 파라미터를 두 방향으로 조금씩 흔드는 과정이다.

## SPSA에서 Cailbration(보정)이란?
=> 하이퍼파라미터(a,c,A)를 실험적으로 조정하여 현재 문제와 하드웨어 환경에서 가장 안정적이고 효율적으로 수렴하도록 만드는 과정이다.

## SPSA의 장점은 적은 측정(2회)으로 Gradient를 추정할 수 있어 Quantum 회로의 비용과 노이즈 문제를 효과적으로 해결한다.

## SPSA의 한계
1. 수렴 속도가 느릴 수 있음
2. Noise에 민감
3. Learning Rate, Perturbation 설정이 민감
4. Plateau(평탄한 영역)에서 비효율
