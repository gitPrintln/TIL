## 큐비트(Qubit)
- QML 에서 데이터를 더풍부하게 데이터를 표현하고 학습하기 위해서, 더 풍부한 Feature space map 형성

-- 머신러닝관점에서 Bit 의 한계?
-> 머신러닝 데이터는 하나의 정답(0 또는 1)이 아니라 여러 가능성이 섞인 확률적 정보로 구성되어 있다.

이처럼 큐비트는 여러 가능성과 복잡한 관계를 더 자연스럽고 풍부하게 표현을 할 수 있다.

## 큐비트란?
- 다양한 양자 시스템을 이용해서 만들고 제어할 수 있는 양자 정보의 기본단위
- 양자(Quantum) + 비트(Bit)
- 동시에 확정적으로 가진다는 것이 아니라 측정전까지 여러 가능성이 함께 존재한다는 뜻
- 측정 후 -> 0 또는 1로 결정

특정: 양자 게이트를 통해 적용하면 상태가 변한다.

초기 상태 0의상태 -> 양자게이트 적용(H 적용)

예시) 날씨 예보
컴퓨터는 0 또는 1, 비옴 또는 비안옴
큐비트는 비안옴(50%) + 비옴(50%) = 100%(측정 후 결정)

큐비트는 Feature Space Generator 의 시작점이다.

QML 에서 큐비트 역할
1. 데이터 표현 단위 2. 가능성 표현단위 3. Feature Space 확장 출발점 4. 변환 대상 5. 측정 대상

-> 큐비트는 q0, q1 각각이 H 를 통해 중첩 , 그 이후 q0, q1 서로를 모르는 상태에서 얽힘을 통해 연관지어짐
